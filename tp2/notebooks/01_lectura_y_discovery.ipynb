## Valores faltantes

## Duplicados

Revisamos duplicados completos y tambien ignorando el id, para ver si hay personas repetidas con distinto identificador.

```python
print('Duplicados completos:', df_train.duplicated().sum())
print('Duplicados sin id   :', df_train.drop(columns=[ID_COL]).duplicated().sum())
```

## Cardinalidad y columnas constantes

Buscamos columnas con un solo valor (no sirven para predecir nada) y columnas muy concentradas en un valor.

```python
card = pd.DataFrame({
    'n_unique': df_train.nunique(),
    'top': df_train.apply(lambda s: s.value_counts(normalize=True).index[0]),
    'top_freq': df_train.apply(lambda s: s.value_counts(normalize=True).iloc[0].round(4)),
})
card.sort_values('n_unique')
```

```python
# Columnas constantes (1 solo valor) -> 'oral' deberia aparecer aca.
constantes = [c for c in df_train.columns if df_train[c].nunique() == 1]
print('Columnas constantes:', constantes)

# Casi-constantes (>95% en un valor).
casi = [c for c in df_train.columns
        if df_train[c].value_counts(normalize=True).iloc[0] > 0.95]
print('Casi-constantes (>95%):', casi)
```
